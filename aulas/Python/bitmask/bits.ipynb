## Bitmask

```python
mask = 0b10110
```

Verificar se um bit tá ligado

```python

for i in range(5):
    if mask & (1 << i):
        print(i, "ligado")
```

Ligando um bit com OR

```python

# ligar o bit 2
mask = mask | (1 << 2)

print(bin(mask))
```

Desligando um bit com AND + NOT


```python
# desligar o bit 1

mask = mask & ~(1<<1)
print(bin(mask))
```

Alternando um bit com XOR

```python
# alternar o bit 3
mask ^= (1 << 3)  
print(bin(mask))
```

Contando e listando bits ligados

```python
qtd = mask.bit_count()
ativos = [i for i in range(5)
            if mask & (1 << i)]
qtd, ativos
```

Gerando todos os subconjuntos


```python
itens = ['A', 'B', 'C']
n = len(itens)

for mask in range(1 << n):
    subconj = []

    for i in range(n):
        if mask & ( 1 << i):
            subconj.append(itens[i])

    print(mask, subconj)
    
```

Filtrando subconjuntos por restrição

```python
pesos = [2, 3, 5, 7]
limite = 8
n = len(pesos)

validos = []
for mask in range(1 << n):
    soma = 0
    for i in range(n):
        if mask & (1 << i):
            soma += pesos[i]

    if soma <= limite:
        validos.append(mask)

print(validos)
```

Sintaxe para percorrer submáscaras

```python
mask = 0b10110

sub = mask
while sub:
    # processa a sub máscara
    print(bin(sub))
    sub = (sub - 1) & mask
```
